```python
# Parser & Space-Fix Utilities
import re
from difflib import SequenceMatcher

ITEM_HEADER_RE = re.compile(
    r"(?:^|\n)\s*(?:#+\s*)?ITEM\s*#?\s*(\d+)\s*:?\s*(?:\n|$)",
    flags=re.IGNORECASE,
)


def parse_segment_score_lines(raw_text):
    """
    Parse lines of the form:
        <segment> -> <score> -> <rank>

    Uses a GREEDY match (.+) to capture the LAST two '->' separators,
    allowing the segment text itself to contain '->'.
    """
    segments = []
    for line in raw_text.strip().splitlines():
        line = line.strip()
        if not line:
            continue

        # GREEDY: (.+) captures everything up to the second-to-last ->
        match = re.match(
            r"^(.+)\s*->\s*([0-9]*\.?[0-9]+)\s*->\s*([0-9]+)\s*$",
            line,
        )
        if match:
            text = match.group(1).rstrip()
            score = float(match.group(2))
            rank = int(match.group(3))

            segments.append({
                "text": text,
                "value_score": score,
                "rank": rank,
            })
    return segments


def split_batch_output(raw_text, n_items):
    """
    Split raw LLM output by ITEM headers into per-item blocks.

    Accepts either 0-based ITEM ids (ITEM 0..N-1) or 1-based ids (ITEM 1..N),
    because models sometimes switch indexing even when the prompt is explicit.
    """
    if n_items <= 0:
        return []

    parts = ITEM_HEADER_RE.split(raw_text)
    raw_blocks = {}
    for i in range(1, len(parts) - 1, 2):
        idx = int(parts[i])
        block = parts[i + 1]
        raw_blocks[idx] = block

    if not raw_blocks:
        if n_items == 1:
            return [parse_segment_score_lines(raw_text)]
        return [[] for _ in range(n_items)]

    keys = set(raw_blocks)
    use_one_based = 0 not in keys and min(keys) >= 1 and max(keys) <= n_items

    item_blocks = {}
    for idx, block in raw_blocks.items():
        normalized_idx = idx - 1 if use_one_based else idx
        if 0 <= normalized_idx < n_items:
            item_blocks[normalized_idx] = block

    result = []
    for i in range(n_items):
        result.append(parse_segment_score_lines(item_blocks.get(i, "")))
    return result


def repair_segments(segments, original_response):
    """
    Repair segments whose text was altered by the LLM.
    Uses the original response as ground truth. Walks through the
    original text greedily, aligning each segment to a substring
    of the original. Gaps between segments are attached to the
    FORMER segment per the space-preservation rule.
    """
    if not segments:
        return segments

    original = original_response.strip()
    repaired = []
    cursor = 0

    for seg in segments:
        seg_text = seg["text"].strip()
        if not seg_text:
            continue

        remaining = original[cursor:]
        if not remaining:
            break

        # 1) Try exact match from cursor
        pos = remaining.find(seg_text)
        if pos != -1:
            gap = remaining[:pos]
            if gap and repaired:
                repaired[-1]["text"] += gap
            elif gap and not repaired:
                seg_text = gap + seg_text
            repaired.append({"text": seg_text, "value_score": seg["value_score"], "rank": seg["rank"]})
            cursor += pos + len(seg_text)
            continue

        # 2) Fuzzy sliding-window match
        best_ratio = 0.0
        best_start = 0
        best_len = len(seg_text)
        search_limit = min(len(remaining), len(seg_text) * 3 + 100)
        for ws in range(max(1, len(seg_text) - 10), len(seg_text) + 15):
            if ws > search_limit:
                break
            for start in range(0, search_limit - ws + 1):
                candidate = remaining[start:start + ws]
                r = SequenceMatcher(None, seg_text, candidate).ratio()
                if r > best_ratio:
                    best_ratio = r
                    best_start = start
                    best_len = ws

        if best_ratio >= 0.55:
            matched = remaining[best_start:best_start + best_len]
            gap = remaining[:best_start]
            if gap and repaired:
                repaired[-1]["text"] += gap
            elif gap and not repaired:
                matched = gap + matched
            repaired.append({"text": matched, "value_score": seg["value_score"], "rank": seg["rank"]})
            cursor += best_start + best_len
        else:
            # Cannot locate - keep as-is
            repaired.append({"text": seg["text"], "value_score": seg["value_score"], "rank": seg["rank"]})

    # Attach any remaining original text to the last segment
    if cursor < len(original) and repaired:
        repaired[-1]["text"] += original[cursor:]

    return repaired


def fix_segment_spaces(segments, original_response):
    """
    Fix space preservation: if concatenating segments drops a space
    that exists in the original, attach it to the FORMER segment.
    """
    if not segments:
        return segments

    original_clean = original_response.strip()
    fixed = [dict(s) for s in segments]

    cursor = 0
    for i, seg in enumerate(fixed):
        seg_text = seg["text"]
        pos = original_clean.find(seg_text, cursor)
        if pos == -1:
            pos = original_clean.find(seg_text.strip(), cursor)
            if pos == -1:
                cursor += len(seg_text)
                continue
            seg_text = seg_text.strip()

        if pos > cursor and i > 0:
            gap = original_clean[cursor:pos]
            fixed[i - 1]["text"] = fixed[i - 1]["text"] + gap

        cursor = pos + len(seg_text)

    return fixed
```

```python
# ── Validation Utilities ──
from difflib import SequenceMatcher

def validate_segments(segments, original_response):
    """
    Validate that concatenated segments faithfully reconstruct the original.
    Returns a dict with validation metrics.
    """
    reconstructed = "".join(seg["text"] for seg in segments)

    # Normalize whitespace for comparison
    orig_norm = " ".join(original_response.split())
    recon_norm = " ".join(reconstructed.split())

    # Exact match (after whitespace normalization)
    exact_match = orig_norm == recon_norm

    # Fuzzy similarity
    similarity = SequenceMatcher(None, orig_norm, recon_norm).ratio()

    # Check for hallucinated words (in reconstruction but not in original)
    orig_words = set(orig_norm.split())
    recon_words = set(recon_norm.split())
    hallucinated_words = recon_words - orig_words
    missing_words = orig_words - recon_words

    # Score sanity checks
    scores = [seg["value_score"] for seg in segments]
    score_issues = []
    for seg in segments:
        s = seg["value_score"]
        if s < 0.0 or s > 1.0:
            score_issues.append(f"Score {s} out of range for: {seg['text'][:40]}")

    # Rank sanity checks
    ranks = [seg["rank"] for seg in segments]
    rank_issues = []
    for seg in segments:
        r = seg["rank"]
        if r < 1 or r > len(segments):
            rank_issues.append(f"Rank {r} out of range for: {seg['text'][:40]}")

    return {
        "exact_match": exact_match,
        "similarity": similarity,
        "n_segments": len(segments),
        "hallucinated_words": hallucinated_words,
        "missing_words": missing_words,
        "score_issues": score_issues,
        "rank_issues": rank_issues,
        "orig_len": len(orig_norm),
        "recon_len": len(recon_norm),
        "reconstructed": reconstructed,
    }


def print_validation_report(validations, verbose=False):
    """Print a summary validation report for a list of validation dicts."""
    print("=" * 70)
    print("VALIDATION REPORT")
    print("=" * 70)
    n = len(validations)
    exact_matches = sum(1 for v in validations if v["exact_match"])
    avg_sim = sum(v["similarity"] for v in validations) / n if n else 0
    total_hallucinated = sum(len(v["hallucinated_words"]) for v in validations)
    total_missing = sum(len(v["missing_words"]) for v in validations)
    total_score_issues = sum(len(v["score_issues"]) for v in validations)
    total_rank_issues = sum(len(v["rank_issues"]) for v in validations)

    print(f"Items validated: {n}")
    print(f"Exact matches (normalized): {exact_matches}/{n} ({100*exact_matches/n:.1f}%)")
    print(f"Average similarity: {avg_sim:.4f}")
    print(f"Total hallucinated words: {total_hallucinated}")
    print(f"Total missing words: {total_missing}")
    print(f"Total score range issues: {total_score_issues}")
    print(f"Total rank range issues: {total_rank_issues}")
    print()

    if verbose:
        for i, v in enumerate(validations):
            if not v["exact_match"]:
                print(f"  Item {i}: similarity={v['similarity']:.4f}")
                if v["hallucinated_words"]:
                    print(f"    Hallucinated: {v['hallucinated_words']}")
                if v["missing_words"]:
                    hw = list(v["missing_words"])[:10]
                    print(f"    Missing (first 10): {hw}")
    print("=" * 70)
```

```python
@kbench.task(name="segment_and_score_batch_fixes")
def segment_and_score_batch(llm, prompts, responses):
    if len(prompts) != len(responses):
        raise ValueError(f"prompts/responses length mismatch: {len(prompts)} != {len(responses)}")

    formatted_inputs = []
    for i, (prompt, response) in enumerate(zip(prompts, responses)):
        formatted_inputs.append(f"""
ITEM {i}

USER PROMPT:
{prompt}

MODEL RESPONSE:
{response}
""")

    batch_input = "\n\n".join(formatted_inputs)

    # UPDATED PROMPT: Aligned with 2026 Process Reward Model (PRM) standards
    prompt_template = POSITIVE_PROMPT if POSITIVE_OR_NEGATIVE else NEGATIVE_PROMPT
    prompt = prompt_template.replace("__BATCH_INPUT__", batch_input)

    # --- LLM Call (Raw text, no schema) ---
    raw_output = None
    last_error = None

    for attempt in range(1, LLM_GENERATION_MAX_RETRIES + 1):
        try:
            raw_output = llm.prompt(
                prompt,
                temperature=LLM_TEMPERATURE,
            )
            break
        except Exception as e:
            last_error = e
            print(f"Batch generation attempt {attempt}/{LLM_GENERATION_MAX_RETRIES} failed: {e}")
            if attempt < LLM_GENERATION_MAX_RETRIES:
                time.sleep(BACKOFF_BASE_SEC * attempt)

    if raw_output is None:
        raise RuntimeError("Max LLM generation retries reached for this batch.") from last_error

    # Convert raw_output to string if needed
    raw_text = str(raw_output)

    # Parse batch output into per-item segments
    parsed_items = split_batch_output(raw_text, len(prompts))
    if len(parsed_items) != len(prompts):
        raise RuntimeError(
            f"Parsed {len(parsed_items)} item(s), expected {len(prompts)}. "
            "Check ITEM headers in the model output."
        )

    # Fix spaces, validate, and store for each item
    for segments, p, response in zip(parsed_items, prompts, responses):
        # Always run fix_segment_spaces as primary correction
        fixed = fix_segment_spaces(segments, response)
        val = validate_segments(fixed, response)

        # If still imperfect, try repair then re-fix spaces
        if not val['exact_match']:
            repaired = repair_segments(segments, response)
            rep_fixed = fix_segment_spaces(repaired, response)
            rep_val = validate_segments(rep_fixed, response)
            if rep_val['similarity'] > val['similarity']:
                fixed = rep_fixed
                val = rep_val

        results.append({
            "segments": fixed,
            "prompt": p,
            "response": response,
            "validation": val,
            "raw_text": raw_text,
        })
```

```python
_DATASET_CACHE = None


def get_dataset():
    global _DATASET_CACHE
    if _DATASET_CACHE is None:
        _DATASET_CACHE = load_dataset(DATASET_NAME)[DATASET_SPLIT]
    return _DATASET_CACHE


def resolve_dataset_range(start_idx, end_idx, dataset=None):
    dataset = dataset if dataset is not None else get_dataset()
    dataset_len = len(dataset)
    start = max(0, min(int(start_idx), dataset_len))
    end = max(start, min(int(end_idx), dataset_len))
    if int(end_idx) > dataset_len:
        print(f"Configured end index {end_idx} exceeds dataset length {dataset_len}; using {end}.")
    if int(start_idx) < 0:
        print(f"Configured start index {start_idx} is below 0; using {start}.")
    return start, end


def collect_batch(start_idx=0, batch_size=BATCH_SIZE, chosen=None, dataset=None):
    dataset = dataset if dataset is not None else get_dataset()
    chosen = POSITIVE_OR_NEGATIVE if chosen is None else chosen

    if batch_size <= 0:
        raise ValueError(f"batch_size must be positive, got {batch_size}")

    start_idx = max(0, min(int(start_idx), len(dataset)))
    stop_idx = min(start_idx + int(batch_size), len(dataset))

    prompts = []
    responses = []
    for idx in range(start_idx, stop_idx):
        row = dataset[idx]
        prompts.append(f"{row['system']}\n{row['question']}")
        responses.append(row["chosen"] if chosen else row["rejected"])

    return prompts, responses, stop_idx


def extract_usage_and_cost(run):
    input_tokens = 0
    output_tokens = 0

    if run.chat is not None:
        for message in run.chat.messages:
            input_tokens += message._meta.get("input_tokens", 0) or 0
            output_tokens += message._meta.get("output_tokens", 0) or 0

    input_cost = (input_tokens / 1_000_000) * PRICING["input_per_mtok_usd"]
    output_cost = (output_tokens / 1_000_000) * PRICING["output_per_mtok_usd"]

    return {
        "input_tokens": input_tokens,
        "output_tokens": output_tokens,
        "input_cost_usd": input_cost,
        "output_cost_usd": output_cost,
        "total_cost_usd": input_cost + output_cost,
    }


def run_single_batch_with_cost(
    start_idx=0,
    batch_size=BATCH_SIZE,
    chosen=None,
    clear_existing_results=True,
    show_results=True,
):
    if clear_existing_results:
        results.clear()

    prompts, responses, stop_idx = collect_batch(
        start_idx=start_idx,
        batch_size=batch_size,
        chosen=chosen,
    )
    run = segment_and_score_batch.run(kbench.llm, prompts, responses)

    if len(results) != len(prompts):
        raise RuntimeError(f"Expected {len(prompts)} result item(s), got {len(results)}.")

    usage = extract_usage_and_cost(run)

    summary = {
        "model": TARGET_LLM,
        "pricing_source_date": PRICING["source_date"],
        "pricing_tier": PRICING.get("pricing_tier"),
        "pricing_source_url": PRICING.get("source_url"),
        "start_idx": start_idx,
        "stop_idx": stop_idx,
        "batch_size": len(prompts),
        **usage,
    }

    print(f"Model: {summary['model']}")
    print(
        f"Batch rows: {summary['start_idx']} to {summary['stop_idx'] - 1} "
        f"({summary['batch_size']} prompts)"
    )
    print(
        f"Tokens: input={summary['input_tokens']:,} | "
        f"output={summary['output_tokens']:,}"
    )
    print(
        f"Estimated cost (USD): input=${summary['input_cost_usd']:.6f} | "
        f"output=${summary['output_cost_usd']:.6f} | "
        f"total=${summary['total_cost_usd']:.6f}"
    )

    # Print validation summary
    validations = [r["validation"] for r in results if "validation" in r]
    if validations:
        print_validation_report(validations, verbose=True)

    if show_results:
        print(f"\nCollected {len(results)} result item(s).")
        print("Segmentation results:")
        show(show_all=True)

    return summary
```

```python
def batch_file_path(start, stop, output_dir=OUTPUT_DIR):
    return f"{output_dir}/orca_segmented_from_{start}_to_{stop}.json"


def failed_indices_file_path(start, stop, output_dir=OUTPUT_DIR):
    return f"{output_dir}/failed_indices_{start}_to_{stop}.json"


def all_failed_indices_path(output_dir=OUTPUT_DIR):
    return f"{output_dir}/all_failed_indices.json"


def failed_batches_path(output_dir=OUTPUT_DIR):
    return f"{output_dir}/failed_batches.json"


def normalize_target_ranges(index_ranges, explicit_indices, dataset=None):
    dataset = dataset if dataset is not None else get_dataset()
    dataset_len = len(dataset)
    normalized = []

    for start_idx, end_idx in index_ranges or []:
        start, end = resolve_dataset_range(start_idx, end_idx, dataset)
        if start < end:
            normalized.append((start, end))

    for idx in explicit_indices or []:
        idx = int(idx)
        if 0 <= idx < dataset_len:
            normalized.append((idx, idx + 1))
        else:
            print(f"Skipping explicit index {idx}; dataset length is {dataset_len}.")

    if not normalized:
        normalized.append(resolve_dataset_range(OVERALL_START, OVERALL_END, dataset))

    normalized.sort()
    merged = []
    for start, end in normalized:
        if not merged or start > merged[-1][1]:
            merged.append([start, end])
        else:
            merged[-1][1] = max(merged[-1][1], end)
    return [(start, end) for start, end in merged]


def build_target_batches(index_ranges, batch_size):
    batches = []
    for range_start, range_end in index_ranges:
        for batch_start in range(range_start, range_end, batch_size):
            batch_stop = min(batch_start + batch_size, range_end)
            batches.append((batch_start, batch_stop))
    return batches


def build_target_index_to_batch(target_batches):
    return {
        idx: (batch_start, batch_stop)
        for batch_start, batch_stop in target_batches
        for idx in range(batch_start, batch_stop)
    }


def export_results(start, stop, output_dir=OUTPUT_DIR):
    expected_count = stop - start
    if len(results) != expected_count:
        raise RuntimeError(
            f"Refusing to export misaligned results: got {len(results)} result item(s), "
            f"expected {expected_count} for rows {start} to {stop - 1}."
        )

    os.makedirs(output_dir, exist_ok=True)
    path = batch_file_path(start, stop, output_dir)
    export_data = []
    failed_indices = []

    for idx, row in enumerate(results):
        item_idx = start + idx
        val = row.get("validation", {})
        export_data.append({
            "index": item_idx,
            "prompt": row["prompt"],
            "response": row["response"],
            "segments": [
                {"text": seg["text"], "value_score": seg["value_score"], "rank": seg["rank"]}
                for seg in row["segments"]
            ],
            "exact_match": val.get("exact_match", False),
            "similarity": val.get("similarity", 0.0),
        })
        if not val.get("exact_match", False):
            failed_indices.append(item_idx)

    with open(path, "w", encoding="utf-8") as f:
        json.dump(export_data, f, indent=2, ensure_ascii=False)
    print(f"Saved {len(export_data)} items to {path}")

    fail_path = failed_indices_file_path(start, stop, output_dir)
    if failed_indices:
        with open(fail_path, "w", encoding="utf-8") as f:
            json.dump(failed_indices, f)
        print(f"  {len(failed_indices)} failed items logged to {fail_path}")
    elif os.path.exists(fail_path):
        os.remove(fail_path)

    return failed_indices
```

```python
dataset = get_dataset()

for job in get_configured_regeneration_jobs():
    POSITIVE_OR_NEGATIVE = bool(job["chosen"])
    TARGET_LABEL_NAME = job["label"]
    TARGET_INDEX_RANGES = job["target_index_ranges"]
    TARGET_EXPLICIT_INDICES = job["target_explicit_indices"]
    OUTPUT_DIR = job["output_dir"]

    target_ranges = normalize_target_ranges(TARGET_INDEX_RANGES, TARGET_EXPLICIT_INDICES, dataset)
    target_batches = build_target_batches(target_ranges, BATCH_SIZE)
    chosen = POSITIVE_OR_NEGATIVE
    os.makedirs(OUTPUT_DIR, exist_ok=True)

    print("=" * 80)
    print(f"Generating {TARGET_LABEL_NAME} side into {OUTPUT_DIR}")
    print(f"Target ranges: {target_ranges[:10]}{' ...' if len(target_ranges) > 10 else ''}")
    print(f"Total target ranges: {len(target_ranges)}")
    print(f"Total target batches: {len(target_batches)}")

    all_failed_indices = []
    persistent_failed_batches = []

    for batch_start, batch_stop in target_batches:
        export_path = batch_file_path(batch_start, batch_stop, OUTPUT_DIR)

        if SKIP_EXISTING_EXPORTS and os.path.exists(export_path):
            print(f"Skipping rows {batch_start} to {batch_stop - 1}; file already exists.")
            existing_fail_path = failed_indices_file_path(batch_start, batch_stop, OUTPUT_DIR)
            if os.path.exists(existing_fail_path):
                with open(existing_fail_path, "r", encoding="utf-8") as f:
                    all_failed_indices.extend(json.load(f))
            continue

        prompts, responses, _ = collect_batch(
            start_idx=batch_start,
            batch_size=batch_stop - batch_start,
            chosen=chosen,
            dataset=dataset,
        )

        if not prompts:
            print(f"No prompts found for range starting at {batch_start}. Skipping.")
            continue

        batch_completed = False

        for attempt in range(1, MAX_RETRIES_PER_BATCH + 1):
            try:
                results.clear()
                segment_and_score_batch.run(kbench.llm, prompts, responses)
                if len(results) != len(prompts):
                    raise RuntimeError(f"Expected {len(prompts)} results, got {len(results)}.")
                failed = export_results(batch_start, batch_stop, OUTPUT_DIR)
                all_failed_indices.extend(failed)
                print(f"Completed rows {batch_start} to {batch_stop - 1} on attempt {attempt}.")
                batch_completed = True
                break
            except Exception as e:
                wait_seconds = BACKOFF_BASE_SEC * attempt
                print(
                    f"Batch {batch_start}-{batch_stop - 1} failed on attempt {attempt}/"
                    f"{MAX_RETRIES_PER_BATCH}: {e}"
                )
                if attempt == MAX_RETRIES_PER_BATCH:
                    print("Max retries reached for this batch; logging and continuing.")
                    persistent_failed_batches.append({"start": batch_start, "stop": batch_stop})
                    if STOP_ON_PERSISTENT_FAILURE:
                        raise
                else:
                    print(f"Sleeping {wait_seconds} seconds before retrying...")
                    time.sleep(wait_seconds)

        if batch_completed:
            time.sleep(SLEEP_BETWEEN_BATCHES_SEC)

    fail_master_path = all_failed_indices_path(OUTPUT_DIR)
    all_failed_indices = sorted(set(int(idx) for idx in all_failed_indices))
    with open(fail_master_path, "w", encoding="utf-8") as f:
        json.dump(all_failed_indices, f)

    failed_batches_out_path = failed_batches_path(OUTPUT_DIR)
    if persistent_failed_batches:
        with open(failed_batches_out_path, "w", encoding="utf-8") as f:
            json.dump(persistent_failed_batches, f, indent=2)
    elif os.path.exists(failed_batches_out_path):
        os.remove(failed_batches_out_path)

    print(f"\n{TARGET_LABEL_NAME}: total failed items after batch pass: {len(all_failed_indices)}")
    print(f"Saved to {fail_master_path}")
    if persistent_failed_batches:
        print(f"Persistent failed batches: {len(persistent_failed_batches)} saved to {failed_batches_out_path}")
```

```python
# Retry Failed Items
# Re-runs ONLY the items that failed validation, one at a time,
# and updates the relevant batch JSON files.

dataset = get_dataset()

for job in get_configured_regeneration_jobs():
    POSITIVE_OR_NEGATIVE = bool(job["chosen"])
    TARGET_LABEL_NAME = job["label"]
    TARGET_INDEX_RANGES = job["target_index_ranges"]
    TARGET_EXPLICIT_INDICES = job["target_explicit_indices"]
    RETRY_OUTPUT_DIR = job["output_dir"]
    retry_chosen = POSITIVE_OR_NEGATIVE

    retry_target_ranges = normalize_target_ranges(TARGET_INDEX_RANGES, TARGET_EXPLICIT_INDICES, dataset)
    retry_target_batches = build_target_batches(retry_target_ranges, RETRY_BATCH_SIZE)
    retry_index_to_batch = build_target_index_to_batch(retry_target_batches)
    os.makedirs(RETRY_OUTPUT_DIR, exist_ok=True)

    print("=" * 80)
    print(f"Retrying failed {TARGET_LABEL_NAME} rows in {RETRY_OUTPUT_DIR}")

    fail_master_path = all_failed_indices_path(RETRY_OUTPUT_DIR)
    failed_batch_path = failed_batches_path(RETRY_OUTPUT_DIR)
    if os.path.exists(fail_master_path):
        with open(fail_master_path, "r", encoding="utf-8") as f:
            failed_indices = json.load(f)
        failed_indices = sorted(set(int(idx) for idx in failed_indices if int(idx) in retry_index_to_batch))
        print(f"Loaded {len(failed_indices)} failed indices in configured target ranges.")
    else:
        failed_indices = []
        print("No failed indices file found.")

    if os.path.exists(failed_batch_path):
        with open(failed_batch_path, "r", encoding="utf-8") as f:
            failed_batches = json.load(f)
        expanded_indices = []
        for batch in failed_batches:
            batch_start = int(batch["start"])
            batch_stop = int(batch["stop"])
            expanded_indices.extend(idx for idx in range(batch_start, batch_stop) if idx in retry_index_to_batch)
        failed_indices = sorted(set(failed_indices).union(expanded_indices))
        print(f"Expanded {len(failed_batches)} failed batch(es) into {len(expanded_indices)} item retry candidate(s).")

    still_failed = []
    fixed_count = 0
    touched_batches = set()

    for item_idx in failed_indices:
        row = dataset[item_idx]
        prompt = f"{row['system']}\n{row['question']}"
        response = row["chosen"] if retry_chosen else row["rejected"]

        best_result = None
        best_sim = 0.0

        for attempt in range(1, RETRY_MAX_ATTEMPTS + 1):
            try:
                results.clear()
                segment_and_score_batch.run(kbench.llm, [prompt], [response])
                if len(results) != 1:
                    raise RuntimeError(f"Expected 1 retry result for idx {item_idx}, got {len(results)}.")

                val = results[0].get("validation", {})
                sim = val.get("similarity", 0)
                if val.get("exact_match", False) or sim > best_sim:
                    best_result = results[0]
                    best_sim = sim
                if val.get("exact_match", False):
                    break
            except Exception as e:
                print(f"  Retry attempt {attempt}/{RETRY_MAX_ATTEMPTS} failed for idx {item_idx}: {e}")

            if attempt < RETRY_MAX_ATTEMPTS:
                time.sleep(RETRY_SLEEP_SEC)

        if best_result is None:
            still_failed.append(item_idx)
            print(f"  idx {item_idx}: ALL RETRIES FAILED")
            continue

        val = best_result.get("validation", {})
        exact = val.get("exact_match", False)

        batch_bounds = retry_index_to_batch.get(item_idx)
        if batch_bounds is None:
            still_failed.append(item_idx)
            print(f"  idx {item_idx}: no configured target batch; skipping update")
            continue
        batch_start, batch_stop = batch_bounds
        batch_path = batch_file_path(batch_start, batch_stop, RETRY_OUTPUT_DIR)
        touched_batches.add((batch_start, batch_stop))

        replacement_entry = {
            "index": item_idx,
            "prompt": best_result["prompt"],
            "response": best_result["response"],
            "segments": [
                {"text": s["text"], "value_score": s["value_score"], "rank": s["rank"]}
                for s in best_result["segments"]
            ],
            "exact_match": exact,
            "similarity": val.get("similarity", 0.0),
        }

        if os.path.exists(batch_path):
            with open(batch_path, "r", encoding="utf-8") as f:
                batch_data = json.load(f)
        else:
            print(f"  idx {item_idx}: missing batch file {batch_path}; creating it with the retry result.")
            batch_data = []

        replaced = False
        for j, entry in enumerate(batch_data):
            if entry.get("index") == item_idx:
                batch_data[j] = replacement_entry
                replaced = True
                break
        if not replaced:
            batch_data.append(replacement_entry)

        batch_data.sort(key=lambda entry: entry.get("index", -1))
        with open(batch_path, "w", encoding="utf-8") as f:
            json.dump(batch_data, f, indent=2, ensure_ascii=False)

        status = "EXACT" if exact else f"sim={val.get('similarity', 0):.4f}"
        if exact:
            fixed_count += 1
        else:
            still_failed.append(item_idx)
        print(f"  idx {item_idx}: {status}")

    still_failed_set = set(still_failed)
    for batch_start, batch_stop in touched_batches:
        batch_failures = [idx for idx in range(batch_start, batch_stop) if idx in still_failed_set]
        batch_fail_path = failed_indices_file_path(batch_start, batch_stop, RETRY_OUTPUT_DIR)
        if batch_failures:
            with open(batch_fail_path, "w", encoding="utf-8") as f:
                json.dump(batch_failures, f)
        elif os.path.exists(batch_fail_path):
            os.remove(batch_fail_path)

    still_failed = sorted(set(still_failed))
    with open(f"{RETRY_OUTPUT_DIR}/still_failed_indices.json", "w", encoding="utf-8") as f:
        json.dump(still_failed, f)
    with open(fail_master_path, "w", encoding="utf-8") as f:
        json.dump(still_failed, f)
    if os.path.exists(failed_batch_path):
        os.remove(failed_batch_path)

    print(f"\n{TARGET_LABEL_NAME}: retry complete: {fixed_count} fixed, {len(still_failed)} still failing.")
    print(f"Still-failed saved to {RETRY_OUTPUT_DIR}/still_failed_indices.json")
    print(f"Updated master failed list at {fail_master_path}")
```

```python
# Combine All Batches into Single JSONL

dataset = get_dataset()

for job in get_configured_regeneration_jobs():
    POSITIVE_OR_NEGATIVE = bool(job["chosen"])
    label_name = job["label"]
    TARGET_INDEX_RANGES = job["target_index_ranges"]
    TARGET_EXPLICIT_INDICES = job["target_explicit_indices"]
    COMBINE_OUTPUT_DIR = job["output_dir"]
    range_slug = job.get("range_slug", "targeted_regeneration_indices")

    combine_target_ranges = normalize_target_ranges(TARGET_INDEX_RANGES, TARGET_EXPLICIT_INDICES, dataset)
    combine_target_batches = build_target_batches(combine_target_ranges, COMBINE_BATCH_SIZE)
    os.makedirs(COMBINE_OUTPUT_DIR, exist_ok=True)

    all_entries = []
    missing_batches = []
    seen_indices = set()

    for batch_start, batch_stop in combine_target_batches:
        batch_path = batch_file_path(batch_start, batch_stop, COMBINE_OUTPUT_DIR)

        if not os.path.exists(batch_path):
            missing_batches.append((batch_start, batch_stop))
            continue

        with open(batch_path, "r", encoding="utf-8") as f:
            batch_data = json.load(f)

        for entry in batch_data:
            idx = entry.get("index")
            if idx in seen_indices:
                print(f"WARNING: duplicate index {idx} in {batch_path}; keeping the first entry.")
                continue
            seen_indices.add(idx)
            all_entries.append(entry)

    all_entries.sort(key=lambda entry: entry.get("index", -1))

    if missing_batches:
        print(f"WARNING: {len(missing_batches)} batch files missing for {label_name}!")
        for s, e in missing_batches[:10]:
            print(f"  Missing: orca_segmented_from_{s}_to_{e}.json")

    jsonl_path = f"{COMBINE_OUTPUT_DIR}/orca_ssr_{label_name}_{range_slug}.jsonl"
    with open(jsonl_path, "w", encoding="utf-8") as f:
        for entry in all_entries:
            out = {
                "index": entry["index"],
                "prompt": entry["prompt"],
                "response": entry["response"],
                "segments": entry["segments"],
                "exact_match": entry.get("exact_match", False),
                "similarity": entry.get("similarity", 0.0),
            }
            f.write(json.dumps(out, ensure_ascii=False) + "\n")

    n_exact = sum(1 for e in all_entries if e.get("exact_match", False))
    n_total = len(all_entries)
    print("=" * 80)
    print(f"Combined {n_total} {label_name} items into {jsonl_path}")
    print(f"Exact matches: {n_exact}/{n_total} ({100*n_exact/max(n_total, 1):.1f}%)")

    json_path = f"{COMBINE_OUTPUT_DIR}/orca_ssr_{label_name}_{range_slug}.json"
    with open(json_path, "w", encoding="utf-8") as f:
        json.dump(all_entries, f, indent=2, ensure_ascii=False)
    print(f"Also saved as {json_path}")
```
